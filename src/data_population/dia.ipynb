```python
import pandas as pd

# import "transcription annotations.csv" to a dataframe
df = pd.read_csv("diarization annotations.csv")
df
```

```python
# for each unique episode in df, cycle through the diarization fields and add them to a list
# then add the list to a dictionary with the episode as the key
diarization_dict = {}
for episode in df["episode"].unique():
    diarization_dict[episode] = None
    for dia in df[df["episode"] == episode]["diarization"]:
        # read in the diarization as a string string dict
        has_empty = False
        dia = eval(dia)
        for key in dia.keys():
            if len(dia[key]) == 0:
                has_empty = True
                break

        # if the diarization is not empty, add it to the list
        if not has_empty and len(dia.keys()) > 0:
            if not diarization_dict.get(episode):
                diarization_dict[episode] = [dia]
            diarization_dict[episode].append(dia)
        else :
            diarization_dict.pop(episode, None)
            

```

```python
data = [
    ("diaz", 1,  {"SPEAKER_01": "Courtney Bursich", "SPEAKER_02": "Courtney Bursich", "SPEAKER_00": "Dr. Josh Axe"}),
    ("P001", 1,  {"SPEAKER_01": "Courtney Versage", "SPEAKER_02": "Courtney Versage", "SPEAKER_00": "Dr. Josh Axe"}),
    ("P002", 1, {"SPEAKER_01": "Courtney Versage", "SPEAKER_02": "Dr. Chris Motley", "SPEAKER_00": " Dr. Josh Axe"}),
    ("diaz", 2, {"SPEAKER_01": "Dr. Josh Axe", "SPEAKER_02": "Dr. Chris Motley", "SPEAKER_00": "Courtney Versage"}),
    ("P003", 2, {"SPEAKER_01": "Courtney Bursich", "SPEAKER_02": "Courtney Bursich", "SPEAKER_00": "Dr. Josh Axe"}),
    ("P004", 2, {"SPEAKER_01": "Courtney Versage", "SPEAKER_02": "Courtney Versage", "SPEAKER_00": "Dr. Axe"}),
    ("P005", 2, {"SPEAKER_01": "Courtney Versage", "SPEAKER_02": "Courtney", "SPEAKER_00": "Dr Axe"}),
    ("P006", 2, {"SPEAKER_01": "Courtney Versage", "SPEAKER_02": "Courtney", "SPEAKER_00": "Dr Axe"}),
]

gold_standard_key = "diaz"  # Key for the gold standard annotations

episodes = set([episode for _, episode, _ in data])

new_data = []

for episode in episodes:
    # Separate gold standard and participant annotations for each episode
    episode_data = [(user_id, annotations) for user_id, episode_id, annotations in data if episode_id == episode]
    gold_standards = [annotations for user_id, annotations in episode_data if user_id == gold_standard_key]
    participant_annotations = [(user_id, annotations) for user_id, annotations in episode_data if user_id != gold_standard_key]

    # Make sure there's only one gold standard
    assert len(gold_standards) == 1
    gold_standard = gold_standards[0]

    for user_id, participant_annotation in participant_annotations:
        new_participant_annotation = {"user_id": user_id, "annotations": [], "is_gold_standard": False}

        for speaker, name in participant_annotation.items():
            # Get the score
            print(f"Gold standard for {speaker} is {gold_standard[speaker]}")
            print(f"{user_id}'s annotation for {speaker} is {name}")
            score = float(input("Enter a score (0-1) for this annotation: "))

            # Save the score
            new_participant_annotation["annotations"].append({speaker: {"name": name, "score": score}})

        new_data.append(new_participant_annotation)
```
